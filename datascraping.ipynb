```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('BTC1min.csv')
df = df.dropna()  # Removes any row with at least one NA value


# 1. Descriptive Statistics
print(df.describe())

# 2. Histograms for Numerical Variables
plt.figure(figsize=(15,10))
numeric_cols = ['Open', 'High', 'Low', 'Close', 'Volume']

for i, col in enumerate(numeric_cols, 1):
    plt.subplot(2, 3, i)
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
plt.tight_layout()
plt.show()

# 3. Box Plots
plt.figure(figsize=(15,6))
sns.boxplot(data=df[numeric_cols])
plt.title('Box Plots of Numerical Variables')
plt.show()

# Step 4: Bivariate Analysis
# 1. Correlation Matrix
plt.figure(figsize=(10,8))
sns.heatmap(df[numeric_cols].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation Matrix')
plt.show()

# 2. Scatter Plots
plt.figure(figsize=(15,10))
plt.subplot(2, 2, 1)
sns.scatterplot(x='Open', y='Close', data=df)
plt.title('Open vs Close')

plt.subplot(2, 2, 2)
sns.scatterplot(x='Volume', y='Close', data=df)
plt.title('Volume vs Close')

plt.subplot(2, 2, 3)
sns.scatterplot(x='High', y='Low', data=df)
plt.title('High vs Low')

plt.tight_layout()
plt.show()

# 3. Line Plot for Price Over Time
plt.figure(figsize=(15,6))
df['Timestamp'] = pd.to_datetime(df['Timestamp'], unit='s')
plt.plot(df['Timestamp'], df['Close'])
plt.title('Bitcoin Closing Price Over Time')
plt.xlabel('Timestamp')
plt.ylabel('Closing Price')
plt.xticks(rotation=45)
plt.show()

```
